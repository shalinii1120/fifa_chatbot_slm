```python
# ==========================================
# CELL 10: DATASET CLEANING
# ==========================================

import pandas as pd
from pathlib import Path

# Project folder
PROJECT_DIR = Path(
    r"C:\Users\nandy\OneDrive\Desktop\final slm project"
)

# Create a separate cleaned copy
fifa_clean = fifa_raw.copy()

# Clean text columns
text_columns = fifa_clean.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    fifa_clean[col] = (
        fifa_clean[col]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
    )

# Remove exact duplicate rows
before_rows = len(fifa_clean)

fifa_clean = fifa_clean.drop_duplicates().reset_index(drop=True)

removed_rows = before_rows - len(fifa_clean)

# Verify essential columns
required_columns = ["player_id", "player_name"]

for col in required_columns:
    if col not in fifa_clean.columns:
        raise ValueError(f"Required column missing: {col}")

if fifa_clean[required_columns].isna().any().any():
    raise ValueError(
        "Some records have missing player IDs or player names. "
        "Inspect these records before continuing."
    )

# Save cleaned dataset
CLEAN_PATH = PROJECT_DIR / "dataset" / "FIFA_CLEANED_DATASET.csv"

CLEAN_PATH.parent.mkdir(parents=True, exist_ok=True)

fifa_clean.to_csv(CLEAN_PATH, index=False)

print("DATASET CLEANING COMPLETED")
print("--------------------------------")
print("Original records:", len(fifa_raw))
print("Cleaned records:", len(fifa_clean))
print("Duplicate rows removed:", removed_rows)
print("Remaining missing values:", fifa_clean.isna().sum().sum())
print("Unique player IDs:", fifa_clean["player_id"].nunique())
print("Unique player names:", fifa_clean["player_name"].nunique())
print("\nSaved at:", CLEAN_PATH)

display(fifa_clean.head())
```

```python
# ==========================================
# CELL 12: PLAYER-LEVEL DATASET SPLITTING
# ==========================================

from sklearn.model_selection import train_test_split

# Get unique player IDs
unique_players = fifa_clean["player_id"].dropna().unique()

print("TOTAL UNIQUE PLAYERS:", len(unique_players))

# First split: 70% training, 30% remaining
train_players, remaining_players = train_test_split(
    unique_players,
    test_size=0.30,
    random_state=42
)

# Second split: remaining 30% into 15% validation and 15% testing
validation_players, test_players = train_test_split(
    remaining_players,
    test_size=0.50,
    random_state=42
)

# Assign all records belonging to each player
train_df = fifa_clean[
    fifa_clean["player_id"].isin(train_players)
].copy()

validation_df = fifa_clean[
    fifa_clean["player_id"].isin(validation_players)
].copy()

test_df = fifa_clean[
    fifa_clean["player_id"].isin(test_players)
].copy()

print("\nSPLITTING COMPLETED")
print("--------------------------------")
print("Training records:", len(train_df))
print("Validation records:", len(validation_df))
print("Testing records:", len(test_df))

print("\nUNIQUE PLAYER COUNTS")
print("Training:", train_df["player_id"].nunique())
print("Validation:", validation_df["player_id"].nunique())
print("Testing:", test_df["player_id"].nunique())

print("\nPERCENTAGE DISTRIBUTION")
total = len(fifa_clean)

print("Training:", round(len(train_df)/total*100, 2), "%")
print("Validation:", round(len(validation_df)/total*100, 2), "%")
print("Testing:", round(len(test_df)/total*100, 2), "%")
```

```python
from sklearn.model_selection import train_test_split
import numpy as np

# Get unique player IDs and convert them to a NumPy array
unique_players = fifa_clean["player_id"].dropna().unique()
unique_players = np.array(unique_players.tolist())

print("TOTAL UNIQUE PLAYERS:", len(unique_players))

# First split: 70% training, 30% remaining
train_players, remaining_players = train_test_split(
    unique_players,
    test_size=0.30,
    random_state=42
)

# Second split: divide remaining 30% equally
validation_players, test_players = train_test_split(
    remaining_players,
    test_size=0.50,
    random_state=42
)

# Create the three datasets using player IDs
train_df = fifa_clean[
    fifa_clean["player_id"].isin(train_players)
].copy()

validation_df = fifa_clean[
    fifa_clean["player_id"].isin(validation_players)
].copy()

test_df = fifa_clean[
    fifa_clean["player_id"].isin(test_players)
].copy()

print("\nSPLIT COMPLETED SUCCESSFULLY!")
print("Training players:", len(train_players))
print("Validation players:", len(validation_players))
print("Testing players:", len(test_players))

print("\nTRAINING ROWS:", len(train_df))
print("VALIDATION ROWS:", len(validation_df))
print("TESTING ROWS:", len(test_df))
```

```python
# Verify that no player appears in multiple datasets

train_ids = set(train_players)
validation_ids = set(validation_players)
test_ids = set(test_players)

print("TRAINING & VALIDATION OVERLAP:",
      len(train_ids.intersection(validation_ids)))

print("TRAINING & TESTING OVERLAP:",
      len(train_ids.intersection(test_ids)))

print("VALIDATION & TESTING OVERLAP:",
      len(validation_ids.intersection(test_ids)))

print("\nTOTAL UNIQUE PLAYERS:",
      len(train_ids | validation_ids | test_ids))

# Confirm that all checks pass
assert len(train_ids.intersection(validation_ids)) == 0
assert len(train_ids.intersection(test_ids)) == 0
assert len(validation_ids.intersection(test_ids)) == 0

assert len(train_ids | validation_ids | test_ids) == len(unique_players)

print("\nALL DATA SPLIT CHECKS PASSED!")
```

```python
# CELL 45: Define FIFA SLM question-answer function

def ask_fifa_model(question):

    inputs = tokenizer(
        question,
        return_tensors="pt",
        truncation=True,
        max_length=128
    )

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        outputs = model.generate(
            **inputs,
            max_new_tokens=64,
            num_beams=4
        )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return answer


print("FIFA SLM FUNCTION CREATED SUCCESSFULLY!")
```

```python
# CELL 56: Evaluate model on 50 unseen test questions

import re

def normalize_answer(text):
    text = str(text).lower().strip()
    text = re.sub(r"\s+", " ", text)
    text = text.strip(" .,!?:;")
    return text

sample_test = test_qa.sample(
    n=min(50, len(test_qa)),
    random_state=42
).reset_index(drop=True)

correct = 0
results = []

for _, row in sample_test.iterrows():

    question = row["input_text"]
    actual = row["target_text"]

    predicted = ask_fifa_model(question)

    is_correct = (
        normalize_answer(predicted) ==
        normalize_answer(actual)
    )

    if is_correct:
        correct += 1

    results.append({
        "Question": question,
        "Actual Answer": actual,
        "Predicted Answer": predicted,
        "Correct": is_correct
    })

accuracy = (correct / len(sample_test)) * 100

print("TOTAL QUESTIONS:", len(sample_test))
print("CORRECT ANSWERS:", correct)
print("INCORRECT ANSWERS:", len(sample_test) - correct)
print("TEST ACCURACY:", round(accuracy, 2), "%")

evaluation_df = pd.DataFrame(results)

display(evaluation_df)
```

```python
# CELL 65: Retrieve player match records

def get_player_matches(player_name):

    result = fifa_clean[
        fifa_clean["player_name"].str.contains(
            player_name,
            case=False,
            na=False,
            regex=False
        )
    ]

    if result.empty:
        print("PLAYER NOT FOUND!")
        return None

    if result["player_id"].nunique() > 1:
        print("Multiple players found. Please specify the exact player.")
        return result[["player_id", "player_name"]].drop_duplicates()

    columns = [
        "player_id",
        "player_name",
        "match_id",
        "match_date",
        "match_result"
    ] + match_stat_columns

    columns = [
        col for col in columns
        if col in result.columns
    ]

    return result[columns].sort_values(
        "match_date"
    )


aaron_matches = get_player_matches("Aaron Goodwin")

display(aaron_matches.head(10))
```

```python
# CELL 76: FIFA question-answer engine

def fifa_question_engine(question):

    player_name = identify_player(question)

    if player_name is None:
        return "PLAYER_NOT_FOUND"

    if player_name == "MULTIPLE_PLAYERS":
        return "MULTIPLE_PLAYERS_FOUND"

    q = question.lower()

    profile = get_player_information(player_name)

    # Player profile questions
    profile_questions = {
        "age": ["age", "old"],
        "nationality": ["nationality"],
        "team": ["national team", "represent"],
        "position": ["position", "play as"],
        "height_cm": ["height"],
        "weight_kg": ["weight"],
        "preferred_foot": ["preferred foot", "strong foot"],
        "club_name": ["club"],
        "market_value_eur": ["market value"]
    }

    for field, keywords in profile_questions.items():

        if any(keyword in q for keyword in keywords):

            if field in profile:
                return f"{player_name}: {profile[field]}"

    # Tournament questions
    tournament_keywords = {
        "total_goals_tournament": ["tournament goals", "total goals"],
        "total_assists_tournament": ["tournament assists", "total assists"],
        "total_minutes_tournament": ["tournament minutes", "total minutes"],
        "player_of_match_awards": ["player of the match awards", "man of the match"],
        "tournament_rating": ["tournament rating"]
    }

    if any(
    keyword in q
    for keywords in tournament_keywords.values()
    for keyword in keywords
):

        summary = get_latest_tournament_summary(player_name)

        for field, keywords in tournament_keywords.items():

            if any(keyword in q for keyword in keywords):
                return f"{player_name}: {summary.get(field, 'Not available')}"

    # Match-specific questions
    match_id = re.search(r"\bM\d+\b", question, re.IGNORECASE)

    if match_id:

        statistics = get_match_statistics(
            player_name,
            match_id.group(0)
        )

        if isinstance(statistics, str):
            return statistics

        statistic_keywords = {
            "goals": ["goals", "scored"],
            "assists": ["assists"],
            "minutes_played": ["minutes"],
            "shots": ["shots"],
            "shots_on_target": ["shots on target"],
            "tackles": ["tackles"],
            "interceptions": ["interceptions"],
            "yellow_cards": ["yellow cards"],
            "red_cards": ["red cards"]
        }

        for field, keywords in statistic_keywords.items():

            if any(keyword in q for keyword in keywords):
                return (
                    f"{player_name} recorded "
                    f"{statistics.get(field, 'Not available')} "
                    f"{field.replace('_', ' ')} in match {match_id.group(0)}."
                )

    return "QUESTION_NOT_SUPPORTED"
```

```python
# CELL 85: Combined FIFA chatbot

def fifa_chatbot(question):

    question = question.strip()

    if not question:
        return "Please enter a question."

    if not is_football_question(question):
        return "Sorry! I can only answer questions related to football and FIFA."

    # First, use the FIFA dataset
    result = fifa_question_engine(question)

    if result not in [
        "PLAYER_NOT_FOUND",
        "QUESTION_NOT_SUPPORTED",
        "MULTIPLE_PLAYERS_FOUND"
    ]:
        return result

    if result == "PLAYER_NOT_FOUND":
        return "This player is not available in our FIFA dataset."

    if result == "MULTIPLE_PLAYERS_FOUND":
        return "Multiple players found. Please specify the player name or ID."

    # Use Ollama for general football questions
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a football knowledge assistant. "
                    "Answer football-related questions clearly. "
                    "Do not invent player statistics or claim "
                    "that information came from the FIFA dataset."
                )
            },
            {
                "role": "user",
                "content": question
            }
        ]
    )

    return response["message"]["content"]
```

```python
# CELL 93: Create FIFA Streamlit chatbot

app_code = r'''
import streamlit as st
import pandas as pd
import ollama
import re
from pathlib import Path

# PROJECT PATHS
ROOT = Path(__file__).resolve().parent.parent

DATASET_PATH = (
    ROOT / "dataset" / "raw" /
    "fifa_world_cup_2026_player_performance.csv"
)

# PAGE CONFIGURATION
st.set_page_config(
    page_title="FIFA Intelligence Chatbot",
    page_icon="⚽",
    layout="wide"
)

st.title("⚽ FIFA Player Intelligence Chatbot")
st.caption("Powered by FIFA Dataset and Ollama Llama 3.2 3B")

# LOAD DATASET
@st.cache_data
def load_dataset():
    return pd.read_csv(DATASET_PATH)

try:
    fifa = load_dataset()
except Exception as e:
    st.error(f"Dataset loading failed: {e}")
    st.stop()

# SIDEBAR
st.sidebar.title("FIFA Assistant")
st.sidebar.success("Dataset connected")
st.sidebar.write(f"Total records: {len(fifa):,}")
st.sidebar.write(
    f"Unique players: {fifa['player_id'].nunique():,}"
)
st.sidebar.write("Ollama model: llama3.2:3b")

# FOOTBALL KEYWORDS
football_keywords = [
    "football", "soccer", "fifa", "world cup",
    "player", "team", "club", "match", "goal",
    "assist", "striker", "defender", "midfielder",
    "goalkeeper", "offside", "penalty", "corner",
    "tackle", "dribble", "passing", "league",
    "coach", "formation", "tournament",
    "footballer", "red card", "yellow card"
]

def is_football_question(question):
    q = question.lower()
    return any(word in q for word in football_keywords)

# PLAYER IDENTIFICATION
def identify_player(question):
    q = question.lower()

    matches = fifa[
        fifa["player_name"].apply(
            lambda name: str(name).lower() in q
        )
    ]

    names = matches["player_name"].dropna().unique()

    if len(names) == 0:
        return None

    if len(names) > 1:
        return "MULTIPLE"

    return names[0]

# PLAYER PROFILE
def get_player_profile(player_name):

    records = fifa[
        fifa["player_name"].str.lower() ==
        player_name.lower()
    ]

    if records.empty:
        return None

    return records.iloc[0]

# FIFA QUESTION ENGINE
def fifa_question_engine(question):

    player_name = identify_player(question)

    if player_name is None:
        return None

    if player_name == "MULTIPLE":
        return "Multiple players found. Please specify the full name."

    q = question.lower()
    player = get_player_profile(player_name)

    # PROFILE QUESTIONS
    profile_fields = {
        "age": ["age", "old"],
        "nationality": ["nationality"],
        "team": ["national team", "represent"],
        "position": ["position", "play as"],
        "height_cm": ["height"],
        "weight_kg": ["weight"],
        "preferred_foot": ["preferred foot"],
        "club_name": ["club"],
        "market_value_eur": ["market value"]
    }

    for field, keywords in profile_fields.items():
        if any(word in q for word in keywords):
            if field in fifa.columns:
                value = player[field]
                if pd.notna(value):
                    return f"{player_name}: {value}"

    # MATCH-SPECIFIC QUESTIONS
    match_found = re.search(
        r"\bM\d+\b", question, re.IGNORECASE
    )

    if match_found:
        match_id = match_found.group(0).upper()

        records = fifa[
            (fifa["player_name"].str.lower() == player_name.lower()) &
            (fifa["match_id"].astype(str).str.upper() == match_id)
        ]

        if records.empty:
            return "No record found for this player in that match."

        if len(records) > 1:
            return "Multiple records found for this player and match."

        record = records.iloc[0]

        stat_keywords = {
            "goals": ["goals", "scored"],
            "assists": ["assists"],
            "minutes_played": ["minutes"],
            "shots": ["shots"],
            "shots_on_target": ["shots on target"],
            "tackles": ["tackles"],
            "interceptions": ["interceptions"],
            "yellow_cards": ["yellow cards"],
            "red_cards": ["red cards"]
        }

        for field, keywords in stat_keywords.items():
            if any(word in q for word in keywords):
                return (
                    f"{player_name} recorded "
                    f"{record[field]} {field.replace('_', ' ')} "
                    f"in match {match_id}."
                )

    return None

# OLLAMA RESPONSE
def ask_ollama(question):

    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a football knowledge assistant. "
                    "Answer only football-related questions. "
                    "Do not invent player statistics."
                )
            },
            {
                "role": "user",
                "content": question
            }
        ]
    )

    return response["message"]["content"]

# CHAT HISTORY
if "messages" not in st.session_state:
    st.session_state.messages = []

for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

# CHAT INPUT
question = st.chat_input("Ask anything about FIFA or football...")

if question:

    st.session_state.messages.append({
        "role": "user",
        "content": question
    })

    with st.chat_message("user"):
        st.markdown(question)

    with st.chat_message("assistant"):

        if not is_football_question(question):
            answer = (
                "Sorry! I can only answer questions "
                "related to football and FIFA."
            )

        else:
            answer = fifa_question_engine(question)

            if answer is None:
                try:
                    answer = ask_ollama(question)
                except Exception as e:
                    answer = (
                        "Ollama connection failed. "
                        "Please check whether Ollama is running. "
                        + str(e)
                    )

        st.markdown(answer)

    st.session_state.messages.append({
        "role": "assistant",
        "content": answer
    })
'''

APP_FILE.write_text(app_code, encoding="utf-8")

print("FIFA CHATBOT APPLICATION CREATED SUCCESSFULLY!")
print("FILE:", APP_FILE)
print("FILE EXISTS:", APP_FILE.exists())
print("FILE SIZE:", APP_FILE.stat().st_size, "bytes")
```
